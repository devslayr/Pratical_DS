## Practical Exercise 3: Basic Operation

```python
fake_abalone =  abalone.copy()
```

```python
## Adding custom columns with values
#Length to height ratio of each row
fake_abalone['Ratio of length to height'] = \
(fake_abalone['Length'] / fake_abalone['Height']).round(4)
```

```python
#fake_abalone['Ratio of length to height']
#abalone['Ratio of length to height']
fake_abalone['Ratio of length to height']
```

```python
# The summation of Shell weight across all rows
fake_abalone['Shell weight'].sum()
```

```python
# Sort data by Length column in descending order
sorted_fake_abalone = \
fake_abalone.sort_values(by='Length', ascending=False)
```

```python
sorted_fake_abalone.head(5)
```

```python
#Drop a single column
# Drop the column named 'Ratio of length to height
sorted_fake_abalone = \
sorted_fake_abalone.drop(columns=['Ratio of length to height'])
```

```python
sorted_fake_abalone.head(5)
```

```python
#Write object to a comma-separated values (csv) file
fake_abalone.to_csv('fake_abalone.csv', index=False)
```

## Practical Exercise 4: Data Preparation

```python
# Add redundant whitespaces
fake_abalone.loc[fake_abalone.Rings.isin([1, 2, 3]), 'Sex'] = 'I '
```

```python
fake_abalone['Sex'].value_counts()
```

```python
#Remove redundant whitespaces
fake_abalone['Sex'] = fake_abalone['Sex'].str.strip()
```

```python
fake_abalone['Sex'].value_counts()
```

```python
#Replace ‘I’ (Infant) labels with ‘K’ (Kid)
fake_abalone['Sex'].replace('I', 'K', inplace=True)
```

```python
#Inject NaN (Not a Number) data
fake_abalone[['Diameter', 'Rings']][fake_abalone.Rings == 22]
```

```python
import numpy as np
fake_abalone.loc[fake_abalone.Rings == 22, 'Diameter'] = np.nan
%rerun 31
```

```python
#Replace missing data with more meaningful values (negative constant, mean and median)
fake_abalone['Diameter'].fillna(fake_abalone['Diameter'].mean(axis=0),
inplace=True)
```
